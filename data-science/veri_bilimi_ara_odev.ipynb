# Veri Bilimi Ara Ödev



```python
import pandas as pd
import numpy as np
```

```python
# 1. Veri setini yükleyin ve ilk 10 satırı ile son 10 satırını ekrana yazdırın.
df = pd.read_csv("e_ticaret_veri_seti.csv") # veri seti yüklendi
```

```python
df.head(10)  # ilk 10 satır
```

```python
df.tail(10) # ilk 10 satır
```

```python
# 2. Veri setinin satır sayısını, sütun sayısını ve tüm sütun isimlerini ekrana yazdırın.
# .shape boyutu verir, .columns.tolist() sütun listesini döndürür.
# .shape[0] satır sayısını, .shape[1] sütun sayısını verir
print(f"Satır Sayısı: {df.shape[0]}")
print(f"Sütun Sayısı: {df.shape[1]}")
print("Sütun İsimleri: \n", df.columns.tolist())
```

```python
# 3. info() ve describe() çıktılarını alın. Ardından veri setindeki sayısal sütunları ve kategorik sütunları ayrı ayrı listeleyin.
print("--- INFO ÇIKTISI ---")
df.info()
print()

print("--- DESCRIBE ÇIKTISI ---")
print(df.describe())
print()

sayisal = df.select_dtypes(include=[np.number]).columns.tolist()
kategorik = df.select_dtypes(include=['object', 'category']).columns.tolist()

print("Sayısal Sütunlar: \n", sayisal)
print("Kategorik Sütunlar: \n", kategorik)
```

```python
# 4. Her sütundaki eksik değer sayısını hesaplayın. Sonra eksik veri oranlarını yüzde olarak çıkarın ve eksik veri oranı en yüksek 5 sütunu yazdırın.
# .isnull().sum() eksik sayıyı, toplam satıra bölünmesi oranını verir. En yüksek 5 değer sıralanır.

eksik_sayi = df.isnull().sum()
eksik_oran = (df.isnull().sum() / len(df)) * 100

eksik_tablo = pd.DataFrame({
    'Eksik Değer Sayısı': eksik_sayi,
    'Eksik Veri Oranı (%)': eksik_oran
})

en_yuksek_5 = eksik_tablo.sort_values(by='Eksik Değer Sayısı', ascending=False).head(5)
print(en_yuksek_5)
```

```python
# 5. En az bir eksik değer içeren satırları filtreleyin ve bu satırlardan ilk 20 tanesini ekrana yazdırın.
# df.isnull().any(axis=1) ile en az bir sütununda NaN olan satırlar çekilir.

eksik_satirlar = df[df.isnull().any(axis=1)]
print(eksik_satirlar.head(20))
```

```python
#6. indirim_orani ve musteri_puani sütunlarındaki eksik değerleri medyan ile doldurun. Doldurmadan önce ve sonra bu iki sütundaki eksik değer sayılarını karşılaştırın.,
# Medyan hesabı aşırı değerlerden etkilenmediği için sayısal eksik verilerde tercih edilir. Fillna sonrası eksik sayı sıfıra düşer.

df['indirim_orani'] = df['indirim_orani'].fillna(df['indirim_orani'].median())
df['musteri_puani'] = df['musteri_puani'].fillna(df['musteri_puani'].median())

print(f"indirim_orani kalan eksik sayısı: {df['indirim_orani'].isnull().sum()}")
print(f"musteri_puani kalan eksik sayısı: {df['musteri_puani'].isnull().sum()}")
```

```python
# 7. odeme_turu ve musteri_tipi sütunlarındaki eksik değerleri mod (en sık geçen değer) ile doldurun.
odeme_mode = df['odeme_turu'].mode()[0]
tip_mode = df['musteri_tipi'].mode()[0]

df['odeme_turu'] = df['odeme_turu'].fillna(odeme_mode)
df['musteri_tipi'] = df['musteri_tipi'].fillna(tip_mode)

print(f"odeme_turu kalan eksik sayısı: {df['odeme_turu'].isnull().sum()}")
print(f"musteri_tipi kalan eksik sayısı: {df['musteri_tipi'].isnull().sum()}")
```

```python
# 8. siparis_tarihi sütununu datetime tipine çevirin. siparis_yili, siparis_ayi, siparis_gunu ve haftanin_gunu sütunlarını türetin.
# Tarih verisi pd.to_datetime ile dönüştürülür ve .dt erişicisi ile zaman bileşenleri ayrıştırılır.
df['siparis_tarihi'] = pd.to_datetime(df['siparis_tarihi'])

df['siparis_yili'] = df['siparis_tarihi'].dt.year
df['siparis_ayi'] = df['siparis_tarihi'].dt.month
df['siparis_gunu'] = df['siparis_tarihi'].dt.day
df['haftanin_gunu'] = df['siparis_tarihi'].dt.day_name()

print(df[['siparis_tarihi', 'siparis_yili', 'siparis_ayi', 'siparis_gunu', 'haftanin_gunu']].head(10))
```

```python
# 9. Veri setindeki tekrarlı(duplicate) satır sayısını bulun ve silin.
# Açıklama: .duplicated().sum() ile tam kopya satır sayısı belirlenir, .drop_duplicates() ile silinir.
tekrarli_satir_sayisi = df.duplicated().sum()
print(f"Tekrarlı Satır Sayısı: {tekrarli_satir_sayisi}")

df = df.drop_duplicates()
print(f"Temizlik Sonrası Satır Sayısı: {len(df)}")
```

```python
# 10. sehir sütunundaki yazım hatalarını ve büyük/küçük harf tutarsızlıklarını standart hale getirin.
# istanbul, ISTANBUL, ıstanbul gibi farklı yazımlar harf dönüşümü ve haritalama (mapping) ile standart Türkçe isim haline getirilir.
def sehir_duzelt(val):
    if pd.isna(val): return val
    val = str(val).strip()
    sozluk = {
        'istanbul': 'İstanbul', 'ISTANBUL': 'İstanbul', 'Istanbul': 'İstanbul', 'ıstanbul': 'İstanbul',
        'izmir': 'İzmir', 'IZMIR': 'İzmir', 'Izmir': 'İzmir',
        'ankara': 'Ankara', 'ANKARA': 'Ankara',
        'bursa': 'Bursa', 'BURSA': 'Bursa',
        'kocaeli': 'Kocaeli', 'KOCAELI': 'Kocaeli',
        'kayseri': 'Kayseri', 'KAYSERI': 'Kayseri',
        'gaziantep': 'Gaziantep', 'GAZIANTEP': 'Gaziantep',
        'antalya': 'Antalya', 'ANTALYA': 'Antalya',
        'konya': 'Konya', 'KONYA': 'Konya'
    }
    return sozluk.get(val, val)

print(f"Temizlik Öncesi Benzersiz Şehir Sayısı: {df['sehir'].nunique(dropna=False)}")
df['sehir'] = df['sehir'].apply(sehir_duzelt)
print(f"Temizlik Sonrası Benzersiz Şehir Sayısı: {df['sehir'].nunique(dropna=False)}")
```

```python
# 11. kategori sütunundaki boşluk, yazım farkı veya küçük/büyük harf sorunlarını temizleyin.
# Kategori isimleri normalize edilir, ana kategoriler eşleştirilerek tutarsızlıklar giderilir.
def kategori_duzelt(val):
    if pd.isna(val): return val
    val = str(val).strip().lower()
    if 'ev' in val and 'yaşam' in val: return 'Ev ve Yaşam'
    elif 'elektronik' in val: return 'Elektronik'
    elif 'kozmet' in val: return 'Kozmetik'
    elif 'kitap' in val: return 'Kitap'
    elif 'spor' in val: return 'Spor'
    elif 'ofis' in val: return 'Ofis'
    return val.title()

df['kategori'] = df['kategori'].apply(kategori_duzelt)
print(df['kategori'].value_counts())
```

```python
# 12. Hatalı/mantıksız verileri tespit edip veri setinden çıkarın (Örn: negatif fiyat, mantıksız puan).
# Negatif birim fiyat ($birim\_fiyat \le 0$), negatif toplam tutar, negatif teslimat günü veya 5'ten büyük müşteri puanları hatalıdır ve veri setinden elenir.
hatali_kosul = (
    (df['birim_fiyat'] <= 0) |
    (df['toplam_tutar'] <= 0) |
    (df['teslimat_gunu'] < 0) |
    (df['musteri_puani'] > 5)
)

hatali_sayisi = hatali_kosul.sum()
print(f"Tespit Edilen Hatalı Satır Sayısı: {hatali_sayisi}")

df_temiz = df[~hatali_kosul].copy()
print(f"Ayıklama Sonrası Kalan Satır Sayısı: {len(df_temiz)}")
```

```python
# 13. birim_fiyat sütunundaki aykırı değerleri (outliers) IQR yöntemi ile tespit edin.
# IQR (Interquartile Range) yöntemi ile $Q1$ (%25) ve $Q3$ (%75) çeyrekleri bulunur. $Q1 - 1.5 \times IQR$ ve $Q3 + 1.5 \times IQR$ sınırlarının dışındaki değerler aykırı değer kabul edilir.
Q1 = df_temiz['birim_fiyat'].quantile(0.25)
Q3 = df_temiz['birim_fiyat'].quantile(0.75)
IQR = Q3 - Q1

alt_sinir = Q1 - 1.5 * IQR
ust_sinir = Q3 + 1.5 * IQR

aykiri_degerler = df_temiz[(df_temiz['birim_fiyat'] < alt_sinir) | (df_temiz['birim_fiyat'] > ust_sinir)]

print(f"Q1: {Q1:.2f}, Q3: {Q3:.2f}, IQR: {IQR:.2f}")
print(f"Alt Sınır: {alt_sinir:.2f}, Üst Sınır: {ust_sinir:.2f}")
print(f"Aykırı Değer Sayısı: {len(aykiri_degerler)}")
```

```python
# 14. Temizlenmiş nihai veri setini yeni bir CSV dosyası olarak kaydedin.
# İşlemleri tamamlanan veri seti index=False parametresi ile yeni bir dosyaya yazdırılır.
df_temiz.to_csv('e_ticaret_temizlenmis_veri_seti.csv', index=False)
print("Temizlenmiş veri seti başarıyla 'e_ticaret_temizlenmis_veri_seti.csv' adıyla kaydedildi!")
```

```python
# 15. Temizlenmiş veri seti üzerinden en az 3 adet temel özet istatistik veya iş içgörüsü çıkarın.
# Veri seti üzerinden ciro, kategori dağılımı ve müşteri davranışlarına dair temel içgörüler elde edilir.
# İçgörü 1: Kategori Bazlı Ortalama Harcama
ort_harcama = df_temiz.groupby('kategori')['toplam_tutar'].mean().sort_values(ascending=False)

# İçgörü 2: Müşteri Tipi Dağılımı ve Ortalama Puanlar
df_temiz['musteri_tipi'] = df_temiz['musteri_tipi'].astype(str).str.strip().str.title().replace({'Vip': 'VIP'})
tip_puan = df_temiz.groupby('musteri_tipi')['musteri_puani'].mean()

# İçgörü 3: Genel Sipariş İstatistikleri
genel_ozet = df_temiz[['birim_fiyat', 'toplam_tutar', 'musteri_puani']].describe()

print("--- 1. Kategoriye Göre Ortalama Sipariş Tutarı ---")
print(ort_harcama)
print()

print("--- 2. Müşteri Tipine Göre Ortalama Müşteri Puanı ---")
print(tip_puan)
print()

print("--- 3. Genel Sayısal Özet İstatistikler ---")
print(genel_ozet)
```
